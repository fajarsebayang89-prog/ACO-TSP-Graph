```python
graph = {
    "A": {"#": 3, "C": 6},
    "#": {"A": 3, "C": 2, "G": 5},
    "C": {"A": 6, "#": 2, "B": 9, "F": 4},
    "B": {"C": 9, "D": 8},
    "D": {"B": 8, "E": 7, "H": 9},
    "F": {"C": 4, "E": 2},
    "E": {"F": 2, "G": 1, "H": 1, "D": 7},
    "G": {"#": 5, "E": 1, "H": 3},
    "H": {"G": 3, "E": 1, "D": 9}
}

print(graph)
```

```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()

edges = [
    ("A", "#", 3),
    ("A", "C", 6),
    ("#", "C", 2),
    ("#", "G", 5),
    ("C", "B", 9),
    ("B", "D", 8),
    ("C", "F", 4),
    ("F", "E", 2),
    ("E", "G", 1),
    ("E", "H", 1),
    ("G", "H", 3),
    ("E", "D", 7),
    ("D", "H", 9)
]

G.add_weighted_edges_from(edges)

pos = nx.spring_layout(G, seed=42)

plt.figure(figsize=(8,6))
nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=2000
)

labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=labels)

plt.show()
```

```python
def hitung_jarak(rute):
    total = 0

    for i in range(len(rute) - 1):
        titik_awal = rute[i]
        titik_tujuan = rute[i + 1]

        if titik_tujuan in graph[titik_awal]:
            total += graph[titik_awal][titik_tujuan]
        else:
            return float("inf")

    return total


rute_uji = ["H", "G", "#", "C", "F", "E", "D"]

jarak = hitung_jarak(rute_uji)

print("Rute:", " -> ".join(rute_uji))
print("Total jarak:", jarak)
```

```python
def cari_semua_jalur(graph, start, goal, path=[]):
    path = path + [start]

    if start == goal:
        return [path]

    paths = []

    for node in graph[start]:
        if node not in path:
            newpaths = cari_semua_jalur(graph, node, goal, path)

            for p in newpaths:
                paths.append(p)

    return paths


semua_jalur = cari_semua_jalur(graph, "H", "D")

print("Semua Jalur H ke D:\n")

for jalur in semua_jalur:
    if "#" in jalur:
        print(jalur)
```

```python
jalur_lewat_pagar = []

for jalur in semua_jalur:
    if "#" in jalur:
        jarak = hitung_jarak(jalur)
        jalur_lewat_pagar.append((jalur, jarak))

print("Jalur H ke D yang melewati #:\n")

for jalur, jarak in jalur_lewat_pagar:
    print(" -> ".join(jalur), "=", jarak)

jalur_terbaik = min(jalur_lewat_pagar, key=lambda x: x[1])

print("\nJalur terbaik:")
print(" -> ".join(jalur_terbaik[0]))
print("Total jarak:", jalur_terbaik[1])
```

```python
import random

start = "H"
goal = "D"
wajib = "#"

jumlah_semut = 100
jumlah_iterasi = 300
alpha = 1
beta = 2
evaporasi = 0.5
Q = 100

pheromone = {}

for node in graph:
    for tetangga in graph[node]:
        pheromone[(node, tetangga)] = 1.0


def pilih_node_sebelah(node, visited):
    pilihan = []

    for tetangga, jarak in graph[node].items():
        if tetangga not in visited:
            tau = pheromone[(node, tetangga)] ** alpha
            eta = (1 / jarak) ** beta
            peluang = tau * eta
            pilihan.append((tetangga, peluang))

    if not pilihan:
        return None

    total = sum(p[1] for p in pilihan)
    r = random.uniform(0, total)

    akumulasi = 0
    for tetangga, peluang in pilihan:
        akumulasi += peluang
        if akumulasi >= r:
            return tetangga

    return pilihan[-1][0]


def buat_rute():
    rute = [start]
    visited = set([start])
    sekarang = start

    while sekarang != goal:
        berikutnya = pilih_node_sebelah(sekarang, visited)

        if berikutnya is None:
            return None

        rute.append(berikutnya)
        visited.add(berikutnya)
        sekarang = berikutnya

    if wajib in rute:
        return rute
    else:
        return None


best_rute = None
best_jarak = float("inf")

for iterasi in range(jumlah_iterasi):
    semua_rute_semut = []

    for semut in range(jumlah_semut):
        rute = buat_rute()

        if rute is not None:
            jarak = hitung_jarak(rute)
            semua_rute_semut.append((rute, jarak))

            if jarak < best_jarak:
                best_jarak = jarak
                best_rute = rute

    for edge in pheromone:
        pheromone[edge] *= (1 - evaporasi)

    for rute, jarak in semua_rute_semut:
        tambahan = Q / jarak

        for i in range(len(rute) - 1):
            a = rute[i]
            b = rute[i + 1]
            pheromone[(a, b)] += tambahan
            pheromone[(b, a)] += tambahan

print("Hasil Ant Colony Optimization")
print("Rute terbaik:", " -> ".join(best_rute))
print("Total jarak:", best_jarak)
```
